# 03 — Feature Adoption

Which features are used and how they relate to retention (Week 2).

**Outputs:** `analysis/outputs/feature_adoption.png`

```python
import os
import sys

REPO_ROOT = os.path.abspath(os.path.join(os.getcwd(), os.pardir)) if os.path.basename(os.getcwd()) == "analysis" else os.getcwd()
if REPO_ROOT not in sys.path:
    sys.path.insert(0, REPO_ROOT)

import asyncio
from analysis.lib.pa_charts import plot_feature_adoption
from db import get_db, init_db
from services import AnalyticsService

OUTPUT_DIR = os.path.join(REPO_ROOT, "analysis", "outputs")
os.makedirs(OUTPUT_DIR, exist_ok=True)
```

```python
async def load_features():
    db_path = os.path.join(REPO_ROOT, "studybuddy.db")
    if not os.path.exists(db_path):
        return {}, {}
    db = await get_db(db_path)
    await init_db(db)
    analytics = AnalyticsService(db)
    usage = await analytics.compute_feature_usage()
    product = await analytics.compute_product_metrics()
    await db.close()
    return usage, product

usage, product = asyncio.run(load_features())
print("Feature retention D7:", product.get("feature_retention", []))
```

```python
features = usage.get("features", [])
if features:
    fig = plot_feature_adoption(features, title="Palph Feature Adoption")
    out = os.path.join(OUTPUT_DIR, "feature_adoption.png")
    fig.savefig(out, dpi=150, bbox_inches="tight")
    print(f"Saved: {out}")
else:
    print("No feature data yet.")
```
